```python
#importing dependencies

import pandas as pd

#mysql toolkit
import pymysql
from sqlalchemy import create_engine
```

```python
print(pd.__version__)
```

```python
df = pd.read_csv('walmart.csv', encoding_errors='ignore')

df.shape
```

```python
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
#all duplicates
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
#dropping all rows with missing records
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df.dtypes
```

```python
df['unit_price'].astype(float)
```

```python
df['unit_price'] = df['unit_price'].str.replace('$','').astype(float)

df.head()
```

```python
df.info()
```

```python
df.columns
```

```python
df['total'] = df['unit_price'] * df['quantity']
df.head()
```

```python
df.to_csv('walmart_clean_data.csv' , index=False)
```

```python
#mysql connection

engine_mysql = create_engine("mysql+pymysql://...:...@.../walmart_db")

try:
    engine_mysql
    print ("Connected to mysql")
except:
    print ("Error")
```
